Program 03: Five-Day Step Tracker

```python
daily_steps = []

for day in range(1, 6):
    steps = int(input(f"Enter steps for day {day}: "))
    daily_steps.append(steps)

total_steps = sum(daily_steps)
highest_steps = max(daily_steps)
average_steps = total_steps / 5

qualifying_days = 0

for steps in daily_steps:
    if steps >= 6000:
        qualifying_days += 1

print("Total steps:", total_steps)
print("Highest daily steps:", highest_steps)
print("Average steps:", round(average_steps, 2))
print("Days with 6000 or more steps:", qualifying_days)
```

Program 04: Stationery Price Manager

```python
prices = {
    "pen": 50,
    "notebook": 180,
    "eraser": 20
}

print("Current stationery prices:")
for item, price in prices.items():
    print(item.title(), ": Rs.", price)

item_name = input("Enter the item you want to update: ").strip().lower() 

if item_name in prices:
    new_price = float(input("Enter the new price: ")) #could be float
    prices.update({item_name: new_price})
    print("Price updated successfully.")
else:
    print("Item not found.")

print("\nUpdated prices:")

for item, price in prices.items():
    print(item.title(), ": Rs.", price)

total_price = sum(prices.values())
print("Total price of all items: Rs.", total_price)
```

Program 05: Three-Subject Result Checker

```python
def average_marks(a: int, b: int, c: int) -> float:
    # first the average of the three marks is calculated and then returned
    return (a + b + c) / 3


# Test the function using assert
assert average_marks(60, 70, 80) == 70.0

mark1 = int(input("Enter marks for subject 1: "))
mark2 = int(input("Enter marks for subject 2: "))
mark3 = int(input("Enter marks for subject 3: "))

average = average_marks(mark1, mark2, mark3)

print("Average marks:", average)

if average >= 50:
    print("Result: Pass")
else:
    print("Result: Needs improvement")
```

Program 06: Visitor Register

```python
visitor1 = input("Enter the first visitor's name: ").strip()
visitor2 = input("Enter the second visitor's name: ").strip()

# Write the first visitor's name to the file
with open("visitors.txt", "w", encoding="utf-8") as file:
    file.write(visitor1 + "\n")

# Add the second visitor's name on a new line
with open("visitors.txt", "a", encoding="utf-8") as file:
    file.write(visitor2 + "\n")

# Read and display the visitors
with open("visitors.txt", "r", encoding="utf-8") as file:
    names = file.read()

print("\nVisitor Register:")
print(names, end="")
```

Program 07: Movie Watchlist with JSON

```python
import json

movies = [
    {"title": "The Lion King", "watched": True},
    {"title": "Finding Nemo", "watched": False},
    {"title": "Coco", "watched": False}
]

# Save movie information in a JSON file in directoy
with open("watchlist.json", "w", encoding="utf-8") as file:
    json.dump(movies, file, indent=4)

# Load the information from the JSON file
with open("watchlist.json", "r", encoding="utf-8") as file:
    loaded_movies = json.load(file)

print("Second movie:", loaded_movies[1]["title"])

print("\nMovies not watched yet:")

for movie in loaded_movies:
    if movie["watched"] == False:
        print(movie["title"])
```

Program 08: Shared-Bill Calculator

```python
try:
    total_bill = float(input("Enter the total restaurant bill: Rs. "))
    friends = int(input("Enter the number of friends sharing the bill: "))

    if friends <= 0:
        print("Please enter a number of friends greater than zero.")
    elif total_bill < 0:
        print("The bill cannot be negative.")
    else:
        amount = total_bill / friends
        print(f"Each person should pay: Rs. {amount:.2f}")

except ValueError:
    print("Invalid input. Please enter a valid number.")
```

Program 09: Parcel Delivery Classes

```python
class Parcel:
    def __init__(self, item_name, weight):
        self.item_name = item_name
        self.weight = weight

    def delivery_fee(self):
        return self.weight * 50


class ExpressParcel(Parcel):
    # Express delivery has a higher fee
    def delivery_fee(self):
        return self.weight * 80


normal_parcel = Parcel("Books", 2)
express_parcel = ExpressParcel("Documents", 2)

print("Normal parcel:", normal_parcel.item_name)
print("Normal delivery fee: Rs.", normal_parcel.delivery_fee())

print("Express parcel:", express_parcel.item_name)
print("Express delivery fee: Rs.", express_parcel.delivery_fee())
```
